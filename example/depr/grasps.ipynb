```python
import json
import pathlib
import pandas as pd
```

```python
paths = pathlib.Path("/workspaces/object_following_ocp/ressources/json/")
```

```python
json_files = list(paths.glob('*.json'))
```

```python
mesh_scale = []
mesh_list = []
scale_list = []
for file in json_files:
    with open(file, "r") as f:
        data = json.load(f)
    for obj in data["objects"].values():
        mesh = obj["mesh"]
        scale = obj["scale"]
        if mesh not in mesh_list:
            mesh_list.append(mesh)
            scale_list.append(scale)
            mesh_scale.append((mesh,scale))
        else:
            print("already in")
            
```

```python
for mesh,scale in mesh_scale:
    print(f"Mesh: {mesh}, Scale: {scale}")
```

```python

df = pd.read_table("/workspaces/object_following_ocp/ressources/PAMI/Grasps.md", sep="|", engine="python", skiprows=[1])
df = df.drop(columns=df.columns[0]).drop(columns=df.columns[-1])  # remove empty edge columns
df.columns = [c.strip() for c in df.columns]

df.to_json("output.json", orient="records", indent=2)
```
